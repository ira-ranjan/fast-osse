```python
from ipyleaflet import Map, Marker, basemaps
import ship_obs_in as ship
import obs_to_obs_seq_in as obsin
import datetime as dt
```

```python
ocean_map = Map(
    center=(0.0, -140.0), 
    zoom=4, 
    basemap=basemaps.Esri.OceanBasemap
)
```

```python
def obs_list_to_markers(obs_list: list):
    coords_list = []
    for obs in obs_list:
        coords = (obs['longitude'],-obs['latitude'])
        if coords not in coords_list:
            coords_list.append(coords)
    for coords in coords_list:
        marker = Marker(location=coords, draggable=False)
        ocean_map.add(marker)
```
