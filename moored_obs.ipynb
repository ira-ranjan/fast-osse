```python
import obs_to_obs_seq_in as obsin
import moored_obs_generator as obsgen
import datetime as dt
```

```python
# Creates MooredObs objects for each of the 5 ADCP-equipped TAO moorings at 140W

taominus1 = obsgen.MooredObs(-1,140,30,250,5,"height (m)",dt.datetime(2017,10,3,12),dt.datetime(2017,12,3,12),0.001,5)
taominus2 = obsgen.MooredObs(-2,140,30,250,5,"height (m)",dt.datetime(2017,10,3,12),dt.datetime(2017,12,3,12),0.001,5)
tao0 = obsgen.MooredObs(0,140,30,250,5,"height (m)",dt.datetime(2017,10,3,12),dt.datetime(2017,10,12,12),0.001,5)
taoplus1 = obsgen.MooredObs(1,140,30,250,5,"height (m)",dt.datetime(2017,10,3,12),dt.datetime(2017,12,3,12),0.001,5)
taoplus2 = obsgen.MooredObs(21,140,30,250,5,"height (m)",dt.datetime(2017,10,3,12),dt.datetime(2017,12,3,12),0.001,5)
```

```python
# Creates an empty obs_sequence object

obs_seq = obsin.create_obs_seq_in()
```

```python
obsin.add_list_to_df(taominus1.metadata, obs_seq)
obsin.add_list_to_df(taominus2.metadata, obs_seq)
obsin.add_list_to_df(tao0.metadata, obs_seq)
obsin.add_list_to_df(taoplus1.metadata, obs_seq)
obsin.add_list_to_df(taoplus2.metadata, obs_seq)
```

```python
print(obs_seq.df)
```

```python
dfs_list = [group for _, group in obs_seq.df.groupby('days')]
for ii in range(len(dfs_list)):
    df_day = dfs_list[ii].to_dict(orient='records')
    obs_seq_day = obsin.create_obs_seq_in()
    obsin.add_list_to_df(df_day,obs_seq_day)
    obs_seq_day.write_obs_seq(f"obs_seq_{dfs_list[ii]['days'].iloc[0]}.test")
```
