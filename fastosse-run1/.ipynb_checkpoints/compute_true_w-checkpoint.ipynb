```python
# My case didn't have volume transports in z-output. Calculate from mass transports
ds['uh'] = ds['umo']/1035.   # kg /s to m^3 /s
ds['vh'] = ds['vmo']/1035. 

# Fix the annoying difference in variable names for the vertical coordinate
ds = ds.rename({'z_l':'zl','z_i':'zi'})
```

```python
ds['true_w'] = ds['vert_remap_h_tendency'].cumsum(dim='zl')
```

```python
def compute_true_w_profile(
    ds, center_lat, center_lon, box_size,
    start_depth, end_depth, depth_step,
    var='true_w', area_weights=None,
):
    """
    Compute a spatial-box-averaged depth profile of true w from a MOM6
    history dataset, interpolated onto an interface-depth grid that matches
    compute_w_planefit's w_est output convention exactly (so the two can be
    subtracted directly, once depth sign is reconciled — see note below).
 
    Parameters
    ----------
    ds : xr.Dataset
        MOM6 history dataset with `var` on dims (time, zl, yh, xh) (or
        equivalent — must have yh/xh as the horizontal dims).
    center_lat, center_lon : float
        Center of the averaging box, in the same units as ds.yh/ds.xh
        (degrees; note xh here runs 190-265, i.e. 0-360 convention —
        make sure center_lon matches that, not a -180/180 convention).
    box_size : float
        Full width/height of the box in degrees (e.g. 1 -> ±0.5 deg).
    start_depth, end_depth, depth_step : float
        Define the MIDPOINT depths conceptually (e.g. 8, 80, 2 means
        obs midpoints at 8,10,...,80). The function returns values at the
        corresponding INTERFACE depths, following the same
        z_top = start - step/2, then step increments, n_midpoints+1 points
        convention compute_w_planefit uses for w_est's depth coordinate.
        For start=8, end=80, step=2 this returns depths [7, 9, ..., 81].
        Depths are returned POSITIVE-DOWN (matching zl's convention) —
        you'll need to negate compute_w_planefit's w_est depth coordinate
        (which is negative-down) before subtracting, not the reverse.
    var : str
        Name of the true-w variable in ds. Default 'w_true'.
    area_weights : xr.DataArray or None
        Optional (yh, xh) cell-area weights for a proper area-weighted mean.
        If None, an UNWEIGHTED mean over the box is used — fine for a small,
        roughly-uniform-area box, but check this assumption if your grid has
        significant latitude-dependent cell area (e.g. areacello in your
        dataset's associated_files) or the box spans a wide latitude range.
 
    Returns
    -------
    xr.DataArray, dims (time, depth), name f'{var}_box_mean', with the
    interpolated, box-averaged profile. `depth` coordinate is positive-down
    interface depths as described above.
    """
    lat_min, lat_max = center_lat - box_size / 2, center_lat + box_size / 2
    lon_min, lon_max = center_lon - box_size / 2, center_lon + box_size / 2
 
    box = ds[var].sel(yh=slice(lat_min, lat_max), xh=slice(lon_min, lon_max))
    if box.sizes.get('yh', 0) == 0 or box.sizes.get('xh', 0) == 0:
        raise ValueError(
            f"Box selection (lat {lat_min}-{lat_max}, lon {lon_min}-{lon_max}) "
            f"returned no grid points — check center_lat/center_lon are in "
            f"the same convention as ds.yh/ds.xh (e.g. 0-360 longitude here, "
            f"not -180/180)."
        )
 

    box_mean = box.mean(dim=['yh', 'xh'])
 
    # --- build the interface-depth grid, following compute_w_planefit's
    # own z_top = obs_z[0] + dz/2 convention, reproduced here so the two
    # grids are numerically identical, not just visually similar.
    n_mid = int(round((end_depth - start_depth) / depth_step)) + 1
    z_top = start_depth - depth_step / 2
    interp_depths = z_top + np.arange(n_mid + 1) * depth_step  # positive-down
 
    zl = box_mean.zl.values
    if interp_depths.max() > zl.max() or interp_depths.min() < zl.min():
        print(f"WARNING: requested depth range ({interp_depths.min()}-"
              f"{interp_depths.max()} m) extends beyond ds's zl range "
              f"({zl.min()}-{zl.max()} m). Out-of-range points will be NaN.")
 
    box_mean_np = box_mean.transpose('time', 'zl').values  # (ntime, nzl)
    out_vals = np.full((box_mean_np.shape[0], len(interp_depths)), np.nan)
    for i in range(box_mean_np.shape[0]):
        out_vals[i] = np.interp(
            interp_depths, zl, box_mean_np[i],
            left=np.nan, right=np.nan,
        )
 
    out = xr.DataArray(
        out_vals, dims=('time', 'depth'),
        coords={'time': box_mean.time, 'depth': interp_depths},
        name=f'{var}_mean',
    )
    out.attrs['center_lat'] = center_lat
    out.attrs['center_lon'] = center_lon
    out.attrs['box_size'] = box_size
    out.attrs['depth_convention'] = 'positive-down, interface levels'

    out.to_netcdf("true_w.nc")
    return out
```

```python
true_w = compute_true_w_profile(ds, 0.5, 220, 1, 8, 80, 2)
```
